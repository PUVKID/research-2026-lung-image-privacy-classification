```python
"""
Privacy Score / SSIM / PSNR comparison
  1) Block conditions
  2) Blur conditions

Style:
  - NanumGothic Bold
  - Text color: black
  - Individual font sizes
  - Legend: upper-right inside graph
  - Legend markers kept
  - Legend lines removed
  - No title
"""

import os
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from matplotlib.lines import Line2D


# ===========================================================================
# NanumGothic Bold
# ===========================================================================

FONT_BOLD = r"fonts\NanumGothicBold.ttf"

if not os.path.exists(FONT_BOLD):
    raise FileNotFoundError(
        f"NanumGothic Bold font not found:\n{FONT_BOLD}"
    )

fm.fontManager.addfont(FONT_BOLD)

FP_BOLD = fm.FontProperties(
    fname=FONT_BOLD
)

print("[font] NanumGothic Bold loaded")
print(FONT_BOLD)


# ===========================================================================
# Data
# ===========================================================================

# ---------------------------------------------------------------------------
# Block
# ---------------------------------------------------------------------------

block_labels = [
    "Original",
    "256",
    "128",
    "64",
    "32",
    "16",
    "8",
    "4",
    "2",
    "1"
]

block_privacy = [
    0.00,
    49.65,
    45.99,
    47.70,
    47.40,
    49.20,
    49.57,
    49.27,
    48.82,
    47.14
]

block_ssim = [
    1.0000,
    0.2407,
    0.3183,
    0.2969,
    0.2752,
    0.2115,
    0.1166,
    0.0358,
    0.0095,
    0.0078
]

block_psnr = [
    100.00,
    6.03,
    7.15,
    7.38,
    7.78,
    7.51,
    7.50,
    7.51,
    7.50,
    7.50
]


# ---------------------------------------------------------------------------
# Blur
# ---------------------------------------------------------------------------

blur_labels = [
    "0",
    "1",
    "2",
    "4",
    "8",
    "16",
    "32",
    "64",
    "128",
    "256"
]

blur_privacy = [
    0.00,
    0.75,
    2.48,
    6.10,
    11.45,
    17.70,
    24.75,
    30.33,
    32.36,
    34.62
]

blur_ssim = [
    1.0000,
    0.8973,
    0.7748,
    0.6578,
    0.5958,
    0.5761,
    0.5635,
    0.5427,
    0.5145,
    0.5033
]

blur_psnr = [
    100.00,
    32.67,
    26.21,
    22.22,
    19.62,
    17.34,
    14.99,
    13.10,
    11.60,
    10.14
]


# ===========================================================================
# Colors
# ===========================================================================

C_PRIVACY = "#08519c"
C_SSIM    = "#6a51a3"
C_PSNR    = "#e6550d"


# ===========================================================================
# Font sizes
# ===========================================================================
#
# 여기 숫자만 따로 수정하면 됨
#
# AXIS_LABEL_FS : X/Y축 제목
# TICK_FS       : X/Y축 눈금
# ANNOT_FS      : Privacy Score 수치
# LEGEND_FS     : 범례 글자
# LEGEND_MS     : 범례 기호
#
# ===========================================================================

AXIS_LABEL_FS = 40
TICK_FS       = 30
ANNOT_FS      = 35
LEGEND_FS     = 30
LEGEND_MS     = 20


# ===========================================================================
# Graph style
# ===========================================================================

LW = 4.2
MS = 13


# ===========================================================================
# Figure function
# ===========================================================================

def make_figure(
    labels,
    privacy,
    ssim,
    psnr,
    x_title,
    out_path
):

    # -----------------------------------------------------------------------
    # Figure
    # -----------------------------------------------------------------------

    fig, ax = plt.subplots(
        figsize=(15, 9.5)
    )

    ax2 = ax.twinx()

    x = list(range(len(labels)))


    # =======================================================================
    # Main graph lines
    # =======================================================================

    # Privacy Score
    ax.plot(
        x,
        privacy,
        color=C_PRIVACY,
        marker="o",
        linewidth=LW,
        markersize=MS,
        markerfacecolor=C_PRIVACY,
        markeredgecolor=C_PRIVACY
    )

    # SSIM
    ax.plot(
        x,
        [s * 100 for s in ssim],
        color=C_SSIM,
        marker="^",
        linewidth=LW,
        markersize=MS,
        linestyle="--",
        markerfacecolor=C_SSIM,
        markeredgecolor=C_SSIM
    )

    # PSNR
    ax2.plot(
        x,
        psnr,
        color=C_PSNR,
        marker="s",
        linewidth=LW,
        markersize=MS,
        linestyle=":",
        markerfacecolor=C_PSNR,
        markeredgecolor=C_PSNR
    )


    # =======================================================================
    # Privacy Score annotations
    # =======================================================================

    for i, (xi, yi) in enumerate(zip(x, privacy)):

        if i % 2 == 0:
            dy = 20
            va = "bottom"
        else:
            dy = -28
            va = "top"

        ax.annotate(
            f"{yi:.1f}",
            (xi, yi),

            textcoords="offset points",
            xytext=(0, dy),

            ha="center",
            va=va,

            fontsize=ANNOT_FS,
            fontproperties=FP_BOLD,
            color="black"
        )


    # =======================================================================
    # X-axis
    # =======================================================================

    ax.set_xticks(x)

    ax.set_xticklabels(
        labels,
        fontsize=TICK_FS,
        fontproperties=FP_BOLD,
        color="black"
    )

    ax.set_xlabel(
        x_title,
        fontsize=AXIS_LABEL_FS,
        fontproperties=FP_BOLD,
        color="black",
        labelpad=16
    )


    # =======================================================================
    # Y-axis
    # =======================================================================

    ax.set_ylabel(
        "Privacy Score / SSIM x100",
        fontsize=AXIS_LABEL_FS,
        fontproperties=FP_BOLD,
        color="black",
        labelpad=16
    )

    ax2.set_ylabel(
        "PSNR (dB, log scale)",
        fontsize=AXIS_LABEL_FS,
        fontproperties=FP_BOLD,
        color="black",
        labelpad=20
    )


    # =======================================================================
    # Axis range
    # =======================================================================

    ax.set_ylim(
        -10,
        112
    )

    ax2.set_yscale(
        "log"
    )

    ax2.set_ylim(
        4,
        140
    )


    # =======================================================================
    # Tick labels
    # =======================================================================

    for label in ax.get_xticklabels():

        label.set_fontproperties(FP_BOLD)
        label.set_fontsize(TICK_FS)
        label.set_color("black")


    for label in ax.get_yticklabels():

        label.set_fontproperties(FP_BOLD)
        label.set_fontsize(TICK_FS)
        label.set_color("black")


    for label in ax2.get_yticklabels():

        label.set_fontproperties(FP_BOLD)
        label.set_fontsize(TICK_FS)
        label.set_color("black")


    # =======================================================================
    # Tick marks
    # =======================================================================

    ax.tick_params(
        axis="x",
        labelsize=TICK_FS,
        colors="black"
    )

    ax.tick_params(
        axis="y",
        labelsize=TICK_FS,
        colors="black"
    )

    ax2.tick_params(
        axis="y",
        labelsize=TICK_FS,
        colors="black"
    )


    # =======================================================================
    # Grid
    # =======================================================================

    ax.grid(
        alpha=0.25
    )


    # =======================================================================
    # Legend
    #
    # ● Privacy Score
    # ▲ Single-scale SSIM x100 (ref.)
    # ■ PSNR, dB (ref.)
    #
    # 기호는 유지
    # 기호를 관통하는 선은 없음
    # =======================================================================

    legend_handles = [

        Line2D(
            [0],
            [0],

            marker="o",
            linestyle="None",

            color=C_PRIVACY,

            markerfacecolor=C_PRIVACY,
            markeredgecolor=C_PRIVACY,

            markersize=LEGEND_MS,

            label="Privacy Score"
        ),

        Line2D(
            [0],
            [0],

            marker="^",
            linestyle="None",

            color=C_SSIM,

            markerfacecolor=C_SSIM,
            markeredgecolor=C_SSIM,

            markersize=LEGEND_MS,

            label="Single-scale SSIM x100 (ref.)"
        ),

        Line2D(
            [0],
            [0],

            marker="s",
            linestyle="None",

            color=C_PSNR,

            markerfacecolor=C_PSNR,
            markeredgecolor=C_PSNR,

            markersize=LEGEND_MS,

            label="PSNR, dB (ref.)"
        )
    ]


    legend = ax.legend(
        handles=legend_handles,

        loc="upper right",

        fontsize=LEGEND_FS,

        frameon=True,
        framealpha=0.92,

        borderpad=0.8,
        labelspacing=0.7,
        handletextpad=0.7,
        borderaxespad=0.8
    )


    # =======================================================================
    # Legend font
    # =======================================================================

    for text in legend.get_texts():

        text.set_fontproperties(FP_BOLD)
        text.set_fontsize(LEGEND_FS)
        text.set_color("black")


    # =======================================================================
    # Legend title / frame
    # =======================================================================

    legend.get_frame().set_linewidth(1.0)
    legend.get_frame().set_edgecolor("black")


    # =======================================================================
    # Layout
    # =======================================================================

    plt.tight_layout()


    # =======================================================================
    # Save
    # =======================================================================

    plt.savefig(
        out_path,
        dpi=300,
        bbox_inches="tight",
        facecolor="white"
    )

    plt.close(fig)

    print(f"[saved] {out_path}")


# ===========================================================================
# Output folder
# ===========================================================================

OUT_DIR = os.path.join(
    os.path.expanduser("~"),
    "Desktop",
    "privacy_score_figures"
)

os.makedirs(
    OUT_DIR,
    exist_ok=True
)


# ===========================================================================
# Generate Block figure
# ===========================================================================

make_figure(
    block_labels,
    block_privacy,
    block_ssim,
    block_psnr,

    "Block Size (px)",

    os.path.join(
        OUT_DIR,
        "privacy_score_block.png"
    )
)


# ===========================================================================
# Generate Blur figure
# ===========================================================================

make_figure(
    blur_labels,
    blur_privacy,
    blur_ssim,
    blur_psnr,

    "Gaussian Blur sigma",

    os.path.join(
        OUT_DIR,
        "privacy_score_blur.png"
    )
)


# ===========================================================================
# Done
# ===========================================================================

print()
print("========================================")
print("All figures saved successfully.")
print(f"Output folder: {OUT_DIR}")
print("========================================")
```
